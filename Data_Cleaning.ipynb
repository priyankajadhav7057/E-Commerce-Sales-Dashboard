```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_excel("Ecommerce_Sales_Dataset.xlsx")

```

```python
print("Dataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
print(df.head())
```

```python
# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
print("\nColumns with Missing Values:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# Clean column names
df.columns = df.columns.str.strip()
```

```python
print(df.columns.tolist())
```

```python
# Remove extra spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)
```

```python
print(df["Order_Date"].dtype)
```

```python
# clean numeric col
numeric_columns = [
    "Quantity",
    "Item_Price",
    "Product_Cost",
    "Rating",
    "Product_Cost",
    "Discount"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

```python
df["Discount"] = (
    df["Discount"]
    .astype(str)
    .str.replace("%", "", regex=False)
)

df["Discount"] = pd.to_numeric(
    df["Discount"],
    errors="coerce"
) / 100
```

```python
print(df["Discount"].head())
```

```python
categorical_columns = [
    "Customer_Country",
    "Product_Category",
    "Product_Name",
    "Platform",
    "Return_Status",
    "Customer_Segment",
    "Region",
    "Payment_Method",
    "Shipping_Mode",
    "Order_Status",
    "Sales_Rep"
]

for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts(dropna=False))
```

```python
df["Platform"] = df["Platform"].str.strip().str.title()

df["Return_Status"] = df["Return_Status"].str.strip().str.title()

df["Order_Status"] = df["Order_Status"].str.strip().str.title()

df["Customer_Segment"] = df["Customer_Segment"].str.strip().str.title()

df["Payment_Method"] = df["Payment_Method"].str.strip().str.title()

df["Shipping_Mode"] = df["Shipping_Mode"].str.strip().str.title()
```

```python
categorical_columns = [
    "Customer_Country",
    "Product_Category",
    "Product_Name",
    "Platform",
    "Return_Status",
    "Customer_Name",
    "Customer_Segment",
    "Region",
    "Payment_Method",
    "Shipping_Mode",
    "Order_Status",
    "Sales_Rep"
]

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(
            df[column].mode()[0]
        )
```

```python
numeric_columns = [
    "Quantity",
    "Item_Price",
    "Product_Cost",
    "Rating"
]

for column in numeric_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(
            df[column].median()
        )
```

```python
print("Invalid Quantity:")
print(df[df["Quantity"] <= 0])
```

```python
print("Invalid Item Price:")
print(df[df["Item_Price"] < 0])

print("\nInvalid Product Cost:")
print(df[df["Product_Cost"] < 0])
```

```python
print("Invalid Ratings:")
print(df[
    (df["Rating"] < 1) |
    (df["Rating"] > 5)
])
```

```python
print("Invalid Discounts:")
print(df[
    (df["Discount"] < 0) |
    (df["Discount"] > 1)
])
```

```python
print(df["Order_Status"].value_counts())
```

```python
print("\n===== FINAL DATA QUALITY CHECK =====")

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv(
    "cleaned_Ecommerce_Sales_Dataset.csv",
    index=False
)

print("\nCleaned dataset saved successfully!")
```
